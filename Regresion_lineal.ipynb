```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy import stats

```

```python
#Variable Independiente-> Poblacion de estudiantes(miles)
x= np.array([2,6,8,8,12,16,20,20,22,25])
#Variable Dependiente-> Ventas Trimestrales(miles de dolares)
y = np.array([58,105,88,118,117,137,157,169,149,202])
#Cantidad de datos
print("numero de datos en x :{} y en y:{}".format(len(x),len(y)))
```

```python
#Suma de los productos de x e y
sum_xy = np.sum(x*y)
#Suma de los cuadrados de x
suma_x2 = np.sum(x**2)
#Suma de x
suma_x = np.sum(x)
#Suma de y
suma_y = np.sum(y)
#numero de muestras
n = len(x)
```

```python
print("Sxy = {},x^2={},Sx{},Sy={}".format(sum_xy,suma_x2,suma_x,suma_y))
```

```python
#Hallar hiperparametros -> b0 y b1
b1 = (n*sum_xy - suma_x*suma_y)/(n*suma_x2 - suma_x**2)
print("El coeficiente b1 = {}".format(b1))
```

```python
b0 = (suma_y/n) - b1*(suma_x/n)
print("El coeficiente b0 = {}".format(b0))
```

## Calculo de las metricas del Modelo

```python
y_pred = b0 + b1*x
```

### COEFICIENTE DE CORRELACION (R)

```python
cov_xy = np.sum((x - np.mean(x)) * (y - np.mean(y)))
std_x = np.sqrt(np.sum((x - np.mean(x))**2))
std_y = np.sqrt(np.sum((y - np.mean(y))**2))

r = cov_xy / (std_x * std_y)

print(f"\nCoeficiente de correlación de Pearson (r) = {r:.4f}")
```

### Coeficiente de determinacion(r^2)

```python
r2 = r**2
print(f"\nCoeficiente de determinación (r^2) = {r2:.4f}")
```

### STC-> Suma total de cuadrados. Varianza Total

```python
stc = np.sum((y - np.mean(y))**2)
print(f"\n Suma total de cuadrados (STC) = {stc:.4f}")
```

### Suma de cuadrados debido a la Regresión (SRC)

```python
scr = np.sum((y_pred - np.mean(y))**2)
print(f"\n Suma de cuadrados debido a la regresión (SCR) = {scr:.4f}")
```

### Suma de Cuadrados del Error (SCE)

```python
sce = np.sum((y - y_pred)**2)
print(f"\n Suma de cuadrados del error (SCE) = {sce:.4f}")
```

### Error estandar de estimacion (SEE)

```python
df = n - 2
```

```python
see = np.sqrt(sce / df)
print(f"\n Error estándar de estimación (SEE) = {see:.4f}")
```

### Predicciones


```python
predict = pd.DataFrame({'x_a_predecir': [12,15,20,26],
                        'y_prediccion': b0 + b1*np.array([12,15,20,26])
                        })
predict
```

### Visualizacion

```python
plt.figure(figsize=(10,6))

#Grafico de dispersión de datos
plt.scatter(x, y, color='red', label='Puntos de datos reales')
#Recta de regresión Lineal
plt.plot(x, y_pred, color='black',linewidth=2, label='Recta de regresión lineal: y = {b0:.2f} + {b1:.2f}x')
plt.title('Regresión Lineal Simple')
plt.xlabel('Variable independiente (x) Poblacion de Estudiantes')
plt.ylabel('Variable dependiente (y) Ventas Trimestrales')
plt.legend()


plt.grid(True)
plt.show
```
